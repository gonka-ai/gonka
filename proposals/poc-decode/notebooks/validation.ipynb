```python
import json
import matplotlib.pyplot as plt

# randomly pick hidden_state indexes for each decode step
HONEST_RUNS_PATH = "../data/validation/20260423T075534Z/artifacts.json"
FRAUD_RUNS_PATH  = "../data/validation/20260423T093355Z/artifacts.json"

HONEST_SETUP = '''
models: RedHatAI/Qwen2.5-7B-Instruct-FP8-dynamic vs RedHatAI/Qwen2.5-7B-Instruct-FP8-dynamic
hardware: NVIDIA RTX 4000 vs A100
'''
FRAUD_SETUP = '''
models: Qwen/Qwen2.5-7B-Instruct-AWQ vs RedHatAI/Qwen2.5-7B-Instruct-FP8-dynamic
hardware: A100 vs A100
'''

def load_mismatches(path):
    with open(path) as f:
        data = json.load(f)
    return [run["n_sphere_mismatches"] for run in data["runs"]]

honest_mismatches = load_mismatches(HONEST_RUNS_PATH)
fraud_mismatches  = load_mismatches(FRAUD_RUNS_PATH)

print(f"Honest runs: {len(honest_mismatches)} examples")
print(f"Fraud runs:  {len(fraud_mismatches)} examples")
print()
print(f"Honest setup: {HONEST_SETUP}")
print(f"Fraud setup:  {FRAUD_SETUP}")

fig, ax = plt.subplots(figsize=(12, 5))

ax.scatter(
    range(len(honest_mismatches)),
    honest_mismatches,
    label="Honest",
    color="steelblue",
    alpha=0.7,
    s=40,
)
ax.scatter(
    range(len(fraud_mismatches)),
    fraud_mismatches,
    label="Fraud",
    color="tomato",
    alpha=0.7,
    s=40,
    marker="x",
)

ax.set_xlabel("Example index")
ax.set_ylabel("n_sphere_mismatches")
ax.set_title("Mismatches per example. \n Randomly picked indexes for hidden_state.")
ax.legend()
ax.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```
