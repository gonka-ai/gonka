# Proposal

We propose to modify sampling in vLLM in the way that $sample(artifact_{i}) = token_{i+1}$ will be reproducible.

Goals: 
- to make sure that logrprobs are transformed using all needed transformaton for sampling type and transform to format which guarantees reproducibility between machines
- make sure top-k from this logprobs is in artifacts in the same format 
- make sure sampling is reproducible on-chain

Further code will show simplified toy example how it can work.

## Validation 

Then, the additional validation to existing procedure which doesn't require GPU or inference:

```python
# Any unused token id to mark low probability token which is not in artifact but was sampled
# Sampler should be modified for sure, to guarantee that
another_token_id = -1

def validate_artifacts(
    generated_tokens: [int],
    artifacts: list[dict[int, Decimal]],
    run_seed: int,
):
    RNG = RNG(run_seed)
    sampler = Sampler(RNG)
    for generated_token, artifact in zip(generated_tokens, artifacts):
        if generated_token not in artifact:
            generated_token = another_token_id
        another_token_prob = Decimal(1) - sum(artifact.values())
        artifact_to_sample = {
            **artifact,
            another_token_id: another_token_prob,
        }
        next_token_id = sampler.sample(artifact_to_sample)
        assert next_token_id == generated_token
```

```python
import numpy as np
import json

# Sampling modification in vLLM in the way that sample(artifact_i) = token_{i+1} will be reproducible:
# - to make sure that logprobs are transformed using all needed transformation for sampling type and transform to format which guarantees reproducibility between machines;
# - make sure top-k from this logprobs is in artifacts in the same format;
# - make sure sampling is reproducible on-chain.

# Logprobs transformations
def normalize_logprobs(logits, temperature=1.0):
    logits = np.array(logits, dtype=np.float64)
    logits = logits / temperature
    exps = np.exp(logits - np.max(logits))
    probs = exps / np.sum(exps)
    logprobs = np.log(probs)
    return np.round(logprobs, 6)

# Top-k artifacts format
def extract_topk(logprobs, k=5):
    topk_idx = np.argsort(logprobs)[-k:][::-1]
    return [{"token_id": int(i), "logprob": float(logprobs[i])} for i in topk_idx]

# Sampling reproducibility
def deterministic_sample(logprobs, seed):
    np.random.seed(seed)
    probs = np.exp(logprobs)
    probs = probs / probs.sum()
    sampled_token = np.random.choice(len(probs), p=probs)
    return int(sampled_token)

def sampling_demo(seed=42, k=5):
    # fixed logits (remove randomness)
    logits = np.array([0.1, 0.4, 0.5, -0.2, 0.0, -0.5, -1.0, 0.3, 0.2, -0.1])
    
    logprobs = normalize_logprobs(logits, temperature=1.0)
    
    topk = extract_topk(logprobs, k=k)
    
    sampled_token = deterministic_sample(logprobs, seed)
    
    artifact = {
        "seed": seed,
        "logprobs": logprobs.tolist(),
        "topk": topk,
        "sampled_token": sampled_token,
    }

    with open("artifact_inference.json", "w") as f:
        json.dump(artifact, f, indent=2)
    
    return artifact


print(json.dumps(sampling_demo(seed=123, k=5), indent=2))
```

```python
import json
import numpy as np

# Check sample(artifact_i) = token_{i+1}.
# Emulates on-chain artifact verification.
def validate_artifact(path="artifact_inference.json"):
    # upload artifact
    with open(path, "r") as f:
        artifact = json.load(f)
    
    seed = artifact["seed"]
    logprobs = np.array(artifact["logprobs"], dtype=np.float64)
    expected_token = artifact["sampled_token"]

    # deterministic sampling
    np.random.seed(seed)
    probs = np.exp(logprobs)
    probs /= probs.sum()
    sampled_token = np.random.choice(len(probs), p=probs)
    
    # compare results
    if int(sampled_token) == int(expected_token):
        print("Validation passed! Sampled token reproducible:", sampled_token)
        return True
    else:
        print("Validation failed!")
        print("Expected:", expected_token, "Got:", sampled_token)
        return False

validate_artifact("artifact_inference.json")
```

```python
import os
import json
import numpy as np

# Creates a chain of inference artifacts (each step has its own seed)
def generate_artifact_chain(num_steps=3, base_seed=100):
    os.makedirs("artifacts", exist_ok=True)
    artifacts = []

    for step in range(num_steps):
        seed = base_seed + step
        artifact = sampling_demo(seed=seed, k=5)
        path = f"artifacts/artifact_step_{step}.json"
        with open(path, "w") as f:
            json.dump(artifact, f, indent=2)
        artifacts.append(path)
        print(f"Saved: {path}")

    return artifacts

# Checks the entire chain of artifacts
def validate_artifact_chain(artifacts):
    all_valid = True
    for path in artifacts:
        print(f"\nValidating {path} ...")
        if not validate_artifact(path):
            all_valid = False
    if all_valid:
        print("\nAll artifacts validated successfully!")
    else:
        print("\nSome artifacts failed validation.")
    return all_valid


artifact_paths = generate_artifact_chain(num_steps=5, base_seed=100)
validate_artifact_chain(artifact_paths)
```
